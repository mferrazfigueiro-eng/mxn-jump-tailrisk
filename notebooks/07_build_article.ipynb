# 7. Rebuild every table, figure and number, then compile the article

`reporting.build_all` writes `paper/tables/*.tex`, `paper/figures/*.pdf` and `paper/numbers.tex`. The manuscript `paper/main.tex` only reads these files, so the PDF always matches the results.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve()
if not (ROOT / 'config.toml').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

import json
import pandas as pd
from IPython.display import display, Image
from mxntail import pipeline, reporting
pd.set_option('display.width', 200); pd.set_option('display.max_columns', 30)
cfg = pipeline.load_config()
RES = pipeline.results_dir()

def cached(stage, output):
    """Run a pipeline stage only if its output file is missing."""
    if not (RES / output).exists():
        print(f'Running stage {stage!r} (this can take a while)...')
        pipeline.STAGES[stage](cfg)
    else:
        print(f'Using cached results/{output}. Delete it to recompute.')
```

```python
reporting.build_all(cfg)
```

```python
import subprocess
r = subprocess.run(['latexmk', '-pdf', '-interaction=nonstopmode', 'main.tex'],
                   cwd=ROOT/'paper', capture_output=True, text=True)
print(r.stdout[-1500:])
```
