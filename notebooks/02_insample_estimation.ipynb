# 2. Full-sample maximum-likelihood estimates (Table 2)

All seven models are estimated on the main sample. Standard errors are robust (sandwich) QMLE standard errors from numerical derivatives. These estimates also define the data-generating processes of the Monte Carlo study (notebook 6).

```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve()
if not (ROOT / 'config.toml').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

import json
import pandas as pd
from IPython.display import display, Image
from mxntail import pipeline, reporting
pd.set_option('display.width', 200); pd.set_option('display.max_columns', 30)
cfg = pipeline.load_config()
RES = pipeline.results_dir()

def cached(stage, output):
    """Run a pipeline stage only if its output file is missing."""
    if not (RES / output).exists():
        print(f'Running stage {stage!r} (this can take a while)...')
        pipeline.STAGES[stage](cfg)
    else:
        print(f'Using cached results/{output}. Delete it to recompute.')
```
